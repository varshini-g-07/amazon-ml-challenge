```python
import pandas as pd

# Adjust filename and separator if needed
df = pd.read_csv('s3://ml-challenge-datasets/6ab10eb3b23ba_student_resource/student_resource/dataset/test/test_source3.tsv', sep='\t')

def detect_script(text):
    if not isinstance(text, str) or not text.strip():
        return 'unknown'
    for ch in text:
        cp = ord(ch)
        if 0x0600 <= cp <= 0x06FF: return 'arabic'
        if 0x0900 <= cp <= 0x097F: return 'devanagari'   # Hindi, Marathi
        if 0x0980 <= cp <= 0x09FF: return 'bengali'
        if 0x0A00 <= cp <= 0x0A7F: return 'gurmukhi'     # Punjabi
        if 0x0A80 <= cp <= 0x0AFF: return 'gujarati'
        if 0x0B00 <= cp <= 0x0B7F: return 'oriya'
        if 0x0B80 <= cp <= 0x0BFF: return 'tamil'
        if 0x0C00 <= cp <= 0x0C7F: return 'telugu'
        if 0x0C80 <= cp <= 0x0CFF: return 'kannada'
        if 0x0D00 <= cp <= 0x0D7F: return 'malayalam'
        if 0x4E00 <= cp <= 0x9FFF: return 'cjk'
        if 0x0400 <= cp <= 0x04FF: return 'cyrillic'
    return 'latin'

name_scripts = df['business_name'].apply(detect_script).value_counts()
addr_scripts = df['business_address'].apply(detect_script).value_counts()

print("business_name scripts:")
print(name_scripts)
print()
print("business_address scripts:")
print(addr_scripts)
```
